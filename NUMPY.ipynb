```python
# Aggregation finctions in numpy
np.sum()
np.max()
np.min()
np.size()
np.mean()
np.cumsum() # cummulative sum
np.cumprod() # cummulative product
```

```python
a = [100,150,199,200,250,130]
b = [10,50,30,40,30,10]
price = np.array(a)
quantity = np.array(b)
c = np.cumprod([price,quantity],axis = 0)
print(c)
print(np.sum(c[1]))
```

```python
# Satatistical function
a = np.array([200,300,150,130,200,280])
print(np.mean(a)) # sum of all the values divide by the numbers of values
```

```python
# -1 represent inversely proportional relationship
# 1 represents proportional relationship
# 0 means no relationship
tabacco_consumption = [30,50,10,30,50,40]
deaths = [100,120,70,100,120,112]
print(np.corrcoef([tabacco_consumption,deaths]))
```

```python
price = [300,100,350,150,200]
sales = [10,20,7,17,3]
print(np.corrcoef([price,sales]))
```

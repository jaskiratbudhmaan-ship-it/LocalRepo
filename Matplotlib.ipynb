#Barplot


```python
import matplotlib.pyplot as plt
y = [98,67,88,95,80]
x = ["Part1","Part2","Part3","Part4","Part5"]
colors =["red","green","blue","pink","orange"]
plt.bar(x,y,color=colors,edgecolor = "black")
plt.xlabel("parts of harry potter",fontsize = 20)
plt.ylabel("Popularity",fontsize = 20)
plt.show()
```

#Line Plot

```python
import matplotlib.pyplot as plt
x = ["day1","day2","day3","day4","day5"]
y = [300,420,250,230,400]
y1 = [500,450,300,250,320]
plt.plot(x,y,marker = "o",ls = ":",color = "pink",label = "week 1")
plt.plot(x,y1,marker = "*",ls = "-",color = "blue",label = "week 2",alpha = 0.5)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
brands = ["oneplus","apple","samsung","nokia","redmi"]
x = [22,35,30,3,10]
c = ["red","silver","blue","pink","orange"]
ex = [0,0,0,0,0.1]
plt.pie(x,labels=brands,colors=c,explode=ex,shadow=True,autopct="%.2f")
plt.show()
```

#Box plot


```python
l = [25,46,57,25,45,13,29,39,46,99,22,12,14,16,55,6,64,52,55,78]
plt.boxplot(l)
plt.show()
```

```python
#lowerframe = minimum value
#upperframe = maximum value
#quartile 1
#quartile 2
# median
# quatile  = 25*(number of elements + 1)/100
# min/lf = q1 - 1.5(IQR)
#max/uf = q2 + 1.5(IQR)
#IQR = q3-q1
```

```python
l1 = [5,6,7,8,8,10,12]
plot_values = [l,l1]
plt.boxplot(plot_values)
plt.show()
```

# Step Plot

```python
from matplotlib import markers
x = ["day1","day2","day3","day4","day5"]
y = [30,40,25,30,40]
plt.step(x,y,marker="*",where="pre")
plt.show()
```

# Stack Plot

```python
days = [1,2,3,4,5,6,7]
NOP1 = [5,10,30,20,35,60,80]
NOP2 = [10,20,30,25,40,50,90]
NOP3 = [8,30,50,60,70,90,100]
plt.stackplot(days,NOP1,NOP2,NOP3,baseline="sym",colors = ["pink","hotpink","magenta"],labels=["week1","week2","week3"])
plt.legend()
plt.show()
```

# MERGE JOIN AND CONCATENATE IN PANDAS

```python
import pandas as pd
```

```python
data1 = {
    "empid" :["e01","e02","e03","e04","e05","e06"],
    "names" :["ram","sham","rahul","vishal","ravi","john"],
    "age"   :[34,56,23,44,32,36]
}
data2 = {
    "empid" :["e01","e07","e03","e04","e08","e06"],
    "salary" :[45000,56000,34000,20000,50000,62000]
}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
```

```python
df1
```

```python
df2
```

```python
#MERGE
print(pd.merge(df1,df2,on = "empid", how = "inner")) #inner default
```

```python
print(pd.merge(df1,df2,on = "empid", how = "left"))
```

```python
print(pd.merge(df1,df2,on = "empid", how = "right"))
```

```python
data1 = {
    "empid" :["e01","e02","e03","e04","e05","e06"],
    "names" :["ram","sham","rahul","vishal","ravi","john"],
    "age"   :[34,56,23,44,32,36]}

data2 = {
    "empid" :["e07","e08","e09","e010","e011","e012"],
    "names" :["ben","nick","james","sevii","tabi","vish"],
    "age"   :[34,56,23,44,32,36]}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
```

```python
print(pd.concat([df1,df2]))
```

# Compare DataFrames

```python
dict = {"Fruits":["mango","apples","banana","papaaya"],
        "price":[100,150,50,35],
        "quantity":[15,10,10,3]}
df = pd.DataFrame(dict)
```

```python
df
```

```python
df2 = df.copy()
df2.loc[0,"price"] = 120
df2.loc[1,"price"] = 175
df2.loc[3,"price"] = 30
df2.loc[0,"quantity"] = 12
df2.loc[1,"quantity"] = 15
df2.loc[3,"quantity"] = 5
```

```python
print(df2)
```

```python
print(df.compare(df2)) #compare
```

```python
print(df.compare(df2,align_axis=0)) # default axis = 1
```

```python
print(df.compare(df2,keep_equal=True))
```

```python
print(df.compare(df2,keep_shape=True))
```

#Pivoting and melting dataframes

```python
dict = {"keys":["k1","k2","k1","k2"],
        "names":["john","ben","david","peter"],
        "house" :["red","blue","green","red"],
        "grades" :[3,8,9,8]
        }
df = pd.DataFrame(dict)
```

```python
df
```

```python
print(df.pivot(index="keys", columns="names", values="house"))
```

```python
print(df.pivot(index="keys", columns="names", values=["house", "grades"]))
```

#Melting

```python
import pandas as pd
dict = {"names":["john","ben","david","peter"],
        "house" :["red","blue","green","red"],
        "grades" :[3,8,9,8]
        }
df = pd.DataFrame(dict)
```

```python
print(pd.melt(df,id_vars= ["names"],value_vars=["house","grades"],var_name="houses & grades",value_name="values"))
```

# Outlier removal using Z-score

```python
import pandas as pd
dataset = pd.read_csv("loan.csv")
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
dataset.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.boxplot(data=dataset,x="CoapplicantIncome")
```

```python
sns.distplot(dataset["CoapplicantIncome"])
plt.show()
```

How to remove outliers directly

```python
# mean-3*standard deviation
min_range = dataset["CoapplicantIncome"].mean()-(3*dataset["CoapplicantIncome"].std())
max_range = dataset["CoapplicantIncome"].mean()+(3*dataset["CoapplicantIncome"].std())
```

```python
new_dataset= dataset[dataset["CoapplicantIncome"]<= max_range]
```

```python
sns.boxplot(data=new_dataset,x="CoapplicantIncome")
plt.show()
```

Using Z-score

```python
import pandas as pd
dataset = pd.read_csv("loan.csv")
dataset.head(3)
```

```python
z_score = (dataset["CoapplicantIncome"]- dataset["CoapplicantIncome"].mean())/(dataset["CoapplicantIncome"].std())
# formula of zscore = x - x(mean)/x(std)
```

```python
dataset["z_score"] = z_score
```

```python
new_d = dataset[dataset["z_score"]<3]
```

```python
new_d
```

```python
sns.boxplot(data=new_d,x="z_score")
plt.show()
```

# Simple Linear Regression

It is mostly used when we have one independent variable in the dataset.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
dataset = pd.read_csv("placement.csv")
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
x = dataset[["cgpa"]]
y = dataset[["package"]]
```

```python
sns.scatterplot(x="cgpa",y="package",data=dataset)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train) # when we fit data it means when the model will be trained it will find the best values of m and c to make the bestfit line
# and accurately predict results
```

Testing

```python
lr.predict([[6.89]])
```

Accuracy Score

```python
lr.score(x_test,y_test)*100
```

If you want to improve the accuracy of the model you need to change the value of random_state

```python
# Best fit values of m(slope) and c(intercept)
lr.coef_ # this returns the coefficient of x present in the line euation y = mx+c
```

```python
lr.intercept_
```

```python
y_prd = lr.predict(x)
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x = "cgpa",y="package",data=dataset,label="original")
plt.plot(dataset["cgpa"],y_prd,c="red",label="prediction line")
plt.legend(["original","prediction line"])
plt.show()
```

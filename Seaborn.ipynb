#Line Plot

```python
import seaborn as sns
import pandas as pd
data = {"days":[1,2,3,4,5],
        "NOP":[50,40,60,30,44]}
df = pd.DataFrame(data)
print(df)
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
data = {"days":[1,2,3,4,5],
        "NOP":[50,40,60,30,44]}
df = pd.DataFrame(data)
# print(df)

sns.lineplot(data=data,x="days",y="NOP")
plt.show()
```

# with dataset

```python
data = pd.read_excel("esd.xlsx")
color = sns.color_palette("virdis")
sns.lineplot(data=data ,x="business unit",y="age",hue="ethinicity",style="gender",palette = color) #hue is used to add more parameters for differentiation
plt.show()
```

# Bar Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = sns.load_dataset("tips")
# print(data)
sns.barplot(data=data,x="day",y="tip",estimator="sum")
plt.plot()
# plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = sns.load_dataset("tips")
# print(data)
sns.barplot(data=data,x="day",y="tip",hue="sex",palette="viridis")
plt.plot()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = sns.load_dataset("tips")
# print(data)
sns.barplot(data=data,x="day",y="tip",hue="sex",palette="spring")
plt.plot()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = sns.load_dataset("tips")
# print(data)
sns.barplot(data=data,x="day",y="tip",hue="sex",palette="GnBu",order =["Sun","Sat","Fri","Thur"]) #order lets us decide in which order we want labels of our x-axis
plt.plot()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = sns.load_dataset("tips")
# print(data)
sns.barplot(data=data,x="day",y="tip",hue="sex",palette="viridis",errorbar=("ci",0))
plt.plot()
#errorbars tell the maximum endpoint our data is going to
#ci refers to confidentiality
```

# Scatter Plot

```python
import pandas as pd
import seaborn as sns
data = sns.load_dataset("tips")
print(data)
```

```python
sns.scatterplot(data=data,x="total_bill",y="tip")
plt.show() #number to number
```

```python
sns.scatterplot(data=data,x="total_bill",y="tip",hue="smoker")
plt.show()
```

```python
sns.scatterplot(data=data,x="total_bill",y="tip",hue="time")
plt.show()
```

```python
sns.scatterplot(data=data,x="total_bill",y="tip",hue="day")
plt.show()
```

```python
sns.scatterplot(data=data,x="total_bill",y="tip",hue="day",size="size") # size represents number of people
plt.show()
```

```python
sns.scatterplot(data=data,x="total_bill",y="tip",hue="day",size="size",marker="*")
plt.show()
```

```python
sns.scatterplot(data=data,x="total_bill",y="tip",hue="day",size="size",marker="D")
plt.show()
```

```python
sns.scatterplot(data=data,x="total_bill",y="tip",hue="day",size="size",marker="o")
plt.show()
```

```python
sns.scatterplot(data=data,x="total_bill",y="tip",hue="day",size="size",marker="o",palette="spring")
plt.show()
```
